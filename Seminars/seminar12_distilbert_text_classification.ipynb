- probabilities should be sorted from largest to smallest.

Useful notes:
- `AutoModelForSequenceClassification.from_pretrained(..., num_labels=...)` creates the classifier head,
- `model(**encoded).logits` gives class logits with shape `[1, num_classes]`,

- `torch.topk(probabilities, k=...)` returns the largest probabilities and their label ids.

## 5. Exercise 4: Attention Map Inspection

Attention maps can be interesting, but be careful: attention is not a complete explanation of the final class decision.

What an attention map shows:

```text
for one layer and one head, how much each query token attends to each key token
```

Task:
- run one text with `output_attentions=True`,
- choose one layer and one head,
- plot the full token-to-token attention matrix,
- optionally compare another head or layer.

Function contract:
- `get_attention_matrix(text, model, tokenizer, layer_index, head_index)` returns `tokens` and `attention_matrix`,
- `attention_matrix` should have shape `[num_tokens, num_tokens]`,
- each row should sum to approximately 1.

How to read the heatmap:
- rows are query tokens: the token doing the attending,
- columns are key tokens: the token being attended to,
- one row answers: where did this token read from?

Useful shape:
- `outputs.attentions[layer_index]` has shape `[batch, num_heads, sequence_length, sequence_length]`.


````python
attention_text = 'Apple announced record revenue after launching its latest chip.'


def get_attention_matrix(text, model, tokenizer, layer_index=0, head_index=0):
    tokens = []
    attention_matrix = None
    return tokens, attention_matrix


def plot_attention_matrix(tokens, attention_matrix, title):
    plt.figure(figsize=(8, 7))
    # Plot attention_matrix as a heatmap here.
    # Put tokens on both axes.
    plt.title(title)
    plt.tight_layout()
    plt.show()


attention_tokens, attention_matrix = get_attention_matrix(
    attention_text,
    model,
    tokenizer,
    layer_index=0,
    head_index=0,
)

# plot_attention_matrix(attention_tokens, attention_matrix, 'Attention map: layer 0, head 0')
print(attention_tokens)
print(attention_matrix)
````

### Checks (Exercise 4)

````python
assert isinstance(attention_tokens, list)
assert len(attention_tokens) > 0
assert attention_matrix is not None

assert attention_matrix.shape[0] == len(attention_tokens)
assert attention_matrix.shape[1] == len(attention_tokens)

row_sums = attention_matrix.sum(axis=1)
for row_sum in row_sums:
    assert abs(float(row_sum) - 1.0) < 0.01

print('Exercise 4 passed.')
````
